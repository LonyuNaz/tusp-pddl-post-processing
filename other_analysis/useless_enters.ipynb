```python
import os
from utils import extract_tfd_plan, extract_tfd_info, count_useless_turns, count_useless_enters, extract_enhsp_plan
```

```python
print('Category 1\n----------')
models = {9:{'useless':0,'total':0}, 17:{'useless':0,'total':0}}
for cfg in models.keys():
    for num_t in range(3, 16):

        plan_file = f"../results/tfd/base1/pln_{cfg}_{num_t}t.txt"
        if not os.path.exists(plan_file):
            continue

        all_plans = extract_tfd_plan(plan_file)
        for plan_lines in all_plans:
            actions, trains, tracks = extract_tfd_info(plan_lines)
            useless_enters, total_enters = count_useless_enters(actions, trains)
            models[cfg]['useless'] += useless_enters
            models[cfg]['total'] += total_enters

print(models)
```
