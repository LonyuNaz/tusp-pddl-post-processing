```python
import os
from utils import extract_tfd_plan, extract_tfd_info, count_useless_turns, count_useless_enters
```

```python
print('Category 1\n----------')
cfgs = [6,7]
for cfg in cfgs:
    for num_t in range(3, 16):

        plan_file = f"category1_tfd/pln_{cfg}_{num_t}t.txt"
        if not os.path.exists(plan_file):
            continue

        plan_lines = extract_tfd_plan(plan_file)
        actions, trains, tracks = extract_tfd_info(plan_lines)
        useless_turns, total_turns = count_useless_turns(actions, trains)
        print(f'Model: {cfg} === Trains: {num_t} --- Useless turns: {useless_turns}/{total_turns}')
```
