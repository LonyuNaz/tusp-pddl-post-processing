```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
x = [(370, 560), (550,980)]

train_schedule = {
    'A':[(0, 10, 'move'), (10,40, 'sevice'), (40,45, 'move')],
    'B':[(10, 20, 'move'), (20,50, 'sevice'), (50,55, 'move')]
}

trains = list(train_schedule.keys())
trains.reverse()

colour_map = {'move':'royalblue', 'sevice':'darkorange'}

fig, ax = plt.subplots(figsize=(6,2))
for train, bars in train_schedule.items():
    i = trains.index(train)
    for x1, x2, typ in bars:
        ax.barh(i,width=x2-x1,left=x1, color=colour_map[typ], label=typ)

ax.set_yticks(range(len(x)))
ax.grid(True, axis='x')
ax.set_xticks([i*10 for i in range(10)])
ax.set_yticklabels([f'train {trains[i]}' for i in range(len(x))])
handles,labels=[],[]
h, l = ax.get_legend_handles_labels()
for i in range(len(h)):
    labell = l[i]
    if labell not in labels:
        handles.append(h[i])
        labels.append(labell)

ax.legend(handles, labels, loc='upper right')
ax.set_title('Temporal planning')
ax.set_xlabel('Time (minutes)')
fig.savefig('modelling_temporal.png')
```
